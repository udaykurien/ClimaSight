```python
import requests
import os
```

```python
OPEN_METEO_BASE_URL = os.environ["OPEN_METEO_BASE_URL"]

params = {
    "latitude": 28.61,
    "longitude":77.21,
    "hourly": "shortwave_radiation,direct_normal_irradiance,diffuse_radiation",
    "past_hours": 24,
    "forecast_hours": 1
}
```

```python
r = requests.get(
    OPEN_METEO_BASE_URL,
    params = params,
    timeout = 30,
)
```

```python
r.raise_for_status()
```

```python
print(r.request.url)
print(r.request.method)
print(r.request.headers)
```

```python
r.json()
```
